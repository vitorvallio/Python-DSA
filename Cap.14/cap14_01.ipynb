## Análise por Regressão Linear Simples com Modelo OLS (Ordinary Least Squares) 

```python
df_dsa.head()
```

```python
y = df_dsa["valor_aluguel"]
```

```python
x = df_dsa["area_m2"]
```

```python
x = sm.add_constant(x)
```

```python
modelo = sm.OLS(y, x)
```

```python
resultado = modelo.fit()
```

```python
print(resultado.summary())
```

## Interpretação dos Resultados do Modelo OLS

```python
plt.figure(figsize = (12, 8))
plt.xlabel("area_m2", size = 16)
plt.ylabel("valor_aluguel", size = 16)
plt.plot(x["area_m2"], y, "o", label = "Dados Reais")
plt.plot(x["area_m2"], resultado.fittedvalues, "r-", label = "Linha de Regressão (Previsões do Modelo)")
plt.legend(loc = "best")
plt.show()
```
