## Expressões Regulares

```python
import re
```

```python
texto = "Meu e-mail é exemplo@gmail.com e você pode me contatar em outro_email@yahoo.com."
```

```python
resultado = len(re.findall("@", texto))
```

```python
print("O caractere '@' apareceu", resultado, "vezes no texto.")
```

```python
resultado = re.findall(r'você (\w+)', texto)
```

```python
print("A palavra após 'você' é: ", resultado[0])
```

```python
emails = re.findall(r'\b[A-za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Z|a-z]{2,}\b', texto)
```

```python
print(emails)
```

```python
text = "O aluno estava incrivelmente perdido, mas encontrou a DSA e rapidamente começpu a aprender."
```

```python
for m in re.finditer(r"\w+mente\b", text):
    print('%02d-%02d: %s' % (m.start(), m.end(), m.group(0)))
```

# REGEX com ChatGPT

```python
musica = '''
Todos os dias quando acordo
Não tenho mais o tempo que passou
Mas tenho muito tempo
Temos todo tempo do mundo
Todos os dias, antes de dormir
Lembro e esqueço como foi o dia
Sempre em frente
Não temos tempo a perder
Nosso suor sagrado
É bem mais belo que esse sangue amargo
E tão sério
E selvagem!
Selvagem!
Selvagem!
Veja o Sol dessa manhã tão cinza
A tempestade que chega é da cor dos teus olhos
Castanhos
Então me abraça forte
Me diz mais uma vez que já estamos
Distantes de tudo
Temos nosso próprio tempo
Temos nosso próprio tempo
Temos nosso próprio tempo
Não tenho medo do escuro
Mas deixe as luzes acesas
Agora
O que foi escondido é o que se escondeu
E o que foi prometido ninguém prometeu
Nem foi tempo perdido
Somos tão jovens
Tão jovens!
Tão jovens!
'''
```

```python
print(musica)
```

```python
# 1- Crie um REGEX para contar quantas vezes o caractere "a" aparece em todo texto da música.

import re

# sua string já definida em musica
# para contar só 'a' minúsculo:
count_a = len(re.findall(r'a', musica))


print("quantas vezes 'a' aparece: ", count_a)  
```

```python
# 2- Crie um REGEX em Python para contar quantas vezes a palavra "tempo" aparece na música.

import re

# regex para palavra inteira 'tempo', ignorando maiúsculas/minúsculas

count_tempo = len(re.findall(r'\btempo\b', musica, flags=re.IGNORECASE))

print(f"A palavra 'tempo' aparece {count_tempo} vezes.")
```

```python
# 3- Crie um REGEX em Python para extrair as palavras seguidas por exclamação.

import re

# regex que captura qualquer sequência de "word characters" (incluindo acentuados) 
# imediatamente seguida de "!"

palavras_com_exclamacao = re.findall(r'\b(\w+)!', musica, flags=re.UNICODE)

print(palavras_com_exclamacao)
```

```python
# 4- Crie um REGEX que extrai qualquer palavra cujo antecessor seja a palavra "esse", 
# e o sucessor seja a palavra "amargo", em um texto.

import re

resultados = re.findall(r'(?<=\besse\s)([^\s]+?)(?=\samargo\b)', musica, flags=re.IGNORECASE)

print("palavra encontrada: ",resultados)  # ex: ['sangue']

```

```python
# 5- Crie um REGEX que retorne as palavras com acento, 
# mas somente os caracteres na palavra que são anteriores ao caracter com acento.

import re

# regex:
#   \b           → início de palavra
#   ([A-Za-z]+?) → grupo 1: uma ou mais letras ASCII (não acentuadas), de forma NÃO-gulosa
#   (?=[\u00C0-\u017F]) 
#                → lookahead: garante que logo à frente há um caractere acentuado

resultados = re.findall(r'\b([A-Za-z]+?)(?=[\u00C0-\u017F])', musica, flags=re.UNICODE)
print(resultados)
```
