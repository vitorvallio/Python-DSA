```python
# Exercício 1 - Crie uma lista de 3 elementos e calcule a terceira potência de cada elemento.

# 1) Definimos uma lista com 3 elementos
lista = [2, 5, 7]

# 2) Calculamos a terceira potência (cubo) de cada elemento
cubos = [x**3 for x in lista]

# 3) Exibimos o resultado
print("Lista original:", lista)
print("Cubos:", cubos)
```

```python
# Exercício 2 - Reescreva o código abaixo, usando a função map(). O resultado final deve ser o mesmo!
palavras = 'A Data Science Academy oferce os melhores cursos de análise de dados do Brasil.'.split()
#resultado = [[w.upper(), w.lower(), len(w)] for w in palavras]
#for i in resultado:
#    print (i)

resultado = list(map(lambda w: [w.upper(), w.lower(), len(w)], palavras))

for i in resultado:
    print(i)
```

```python
# Exercício 3 - Calcule a matriz transposta da matriz abaixo.
# Caso não saiba o que é matriz transposta, visite este link: https://pt.wikipedia.org/wiki/Matriz_transposta
# Matriz transposta é um conceito fundamental na construção de redes neurais artificiais, base de sistemas de IA.
matrix = [[1, 2],
          [3, 4],
          [5, 6],
          [7, 8]]

# usando zip(*) para transpor
transposta = [list(coluna) for coluna in zip(*matrix)]

print(transposta)
```

```python
# Exercício 4 - Crie duas funções, uma para elevar um número ao quadrado e outra para elevar ao cubo. 
# Aplique as duas funções aos elementos da lista abaixo. 
# Obs: as duas funções devem ser aplicadas simultaneamente.
lista = [0, 1, 2, 3, 4]

# 1) Definindo as funções
def quadrado(x):
    return x ** 2

def cubo(x):
    return x ** 3

# 2) Lista de entrada
lista = [0, 1, 2, 3, 4]

# 3) Aplicando as duas funções em paralelo e combinando os resultados
#    map(quadrado, lista) → gera todos os quadrados
#    map(cubo, lista)      → gera todos os cubos
#    zip(...)              → emparelha quadrado e cubo de cada elemento
resultados = list(zip(
    map(quadrado, lista),
    map(cubo, lista)
))

print(resultados)
```

```python
# Exercício 5 - Abaixo você encontra duas listas. Faça com que cada elemento da listaA seja elevado 
# ao elemento correspondente na listaB.
listaA = [2, 3, 4]
listaB = [10, 11, 12]

resultado = list(map(pow, listaA, listaB))
print(resultado)
```

```python
# Exercício 6 - Considerando o range de valores abaixo, use a função filter() para retornar 
# apenas os valores negativos.
valores = range(-5, 5)
# Filtrando apenas os negativos

negativos = list(filter(lambda x: x < 0, valores))

print(negativos)
```

```python
# Exercício 7 - Usando a função filter(), encontre os valores que são comuns às duas listas abaixo.
a = [1,2,3,5,7,9]
b = [2,3,5,6,7,8]

# Usando filter() com lambda:
comuns = list(filter(lambda x: x in b, a))

print(comuns)
```

```python
# Exercício 8 - Considere os dois dicionários abaixo. 
# Crie um terceiro dicionário com as chaves do dicionário 1 e os valores do dicionário 2.
dict1 = {'a':1,'b':2}
dict2 = {'c':4,'d':5}

# zip das keys de dict1 com os values de dict2
dict3 = dict(zip(dict1.keys(), dict2.values()))

print(dict3)
```

```python
# Exercício 9 - Considere a lista abaixo e retorne apenas os elementos cujo índice for maior que 5.
lista = ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h']
resultado = lista[6:]  # pega do índice 6 em diante
print(resultado)
```

```python
# Exercício 10 - Crie um regex em Python para extrair a palavra que aparece depois das palavras 
# Data e Science na frase: 'A Data Science Academy oferce os melhores cursos de análise de dados do Brasil.'
import re
texto = 'A Data Science Academy oferece os melhores cursos de análise de dados do Brasil.'

# Captura a palavra após "Data Science"
pattern = r'\bData\s+Science\s+(\w+)'

match = re.search(pattern, texto)
if match:
    print(match.group(1))
```
