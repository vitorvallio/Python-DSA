## Função Map

```python
def potencia(x):
    return x ** 2
```

```python
numeros = [1,2,3,4,5]
```

```python
numeros_ao_quadrado = list(map(potencia, numeros))
```

```python
print(numeros_ao_quadrado)
```

```python
def fahenheit(T):
    return((float(9)/5)*T+32)

def celsius(T):
    return(float(5)/(9)*(T-32))
```

```python
temperaturas = [0,22.5,40,100]
```

```python
map(fahenheit, temperaturas)
```

```python
list(map(fahenheit, temperaturas))
```

```python
for temp in map(fahenheit, temperaturas):
    print(temp)
```

```python
map(celsius, temperaturas)
```

```python
list(map(celsius, temperaturas))
```

```python
map(lambda x: (5.0/9)*(x-32), temperaturas)
```

```python
list(map(lambda x: (5.0/9)*(x-32), temperaturas))
```

```python
a = [1,2,3,4]
b = [5,6,7,8]
```

```python
list(map(lambda x,y:x+y, a,b))
```

```python
a = [1,2,3,4]
b = [5,6,7,8]
c = [9,10,11,12]
```

```python
list(map(lambda x, y, z : x+y+z, a,b,c))
```
