## Função Zip

```python
x = [1,2,3]
y = [4,5,6]
```

```python
zip(x,y)
```

```python
list(zip(x,y))
```

```python
list(zip('ABCD', 'xy'))
```

```python
a = [1,2,3]
b = [4,5,6,7,8]
```

```python
list(zip(a,b))
```

```python
d1 = {'a':1, 'b':2}
d2 = {'c':4, 'd':5}
```

```python
list(zip(d1,d2.values()))
```

```python
def trocaValores(d1, d2):
    dicTemp = {}
    for d1key, d2val in zip(d1, d2.values()):
        dicTemp[d1key] = d2val

    return dicTemp
```

```python
trocaValores(d1,d2)
```
