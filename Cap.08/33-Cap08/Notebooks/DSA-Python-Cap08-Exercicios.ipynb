```python
# Exercício 1 - Crie um objeto a partir da classe abaixo, chamado roc1, 
# passando 2 parâmetros e depois faça uma chamada aos atributos e métodos
from math import sqrt

class Rocket():
     
    def __init__(self, x=0, y=0):
        self.x = x
        self.y = y
        
    def move_rocket(self, x_increment=0, y_increment=1):
        self.x += x_increment
        self.y += y_increment
        
    def print_rocket(self):
        print(self.x, self.y)
        


# 1) Criando o objeto roc1, passando x=5 e y=10
roc1 = Rocket(5, 10)

# 2) Acessando diretamente os atributos
print("Posição inicial:")
print("  x =", roc1.x)
print("  y =", roc1.y)

# 3) Chamando o método move_rocket para deslocar a rocket
#    aqui vamos incrementar x em +3 e y em +2
roc1.move_rocket(x_increment=3, y_increment=2)

# 4) Imprimindo a posição atual usando o método print_rocket
print("\nApós mover a rocket:")
roc1.print_rocket()
```

```python
# Exercício 2 - Crie uma classe chamada Pessoa() com os atributos: nome, cidade, telefone e e-mail. Use pelo menos 2
# métodos especiais na sua classe. Crie um objeto da sua classe e faça uma chamada a pelo menos um dos seus métodos
# especiais.

class Pessoa:
    def __init__(self, nome, cidade, telefone, email):
        self.nome = nome
        self.cidade = cidade
        self.telefone = telefone
        self.email = email

    def __repr__(self):
        # representação “oficial” (útil para debugging)
        return (f"Pessoa(nome={self.nome!r}, cidade={self.cidade!r}, "
                f"telefone={self.telefone!r}, email={self.email!r})")

    def __str__(self):
        # representação “amigável” (quando você faz print(obj))
        return (f"{self.nome} mora em {self.cidade}.\n"
                f"Telefone: {self.telefone} | E-mail: {self.email}")

    def __eq__(self, outro):
        # define igualdade: duas pessoas são “iguais” se tiverem
        # o mesmo e-mail
        if not isinstance(outro, Pessoa):
            return False
        return self.email == outro.email

    def __len__(self):
        # retorna “comprimento” do objeto: número de atributos
        return 4
        

# Criando um objeto da classe
p = Pessoa(
    nome="Vitor Vallio",
    cidade="São Paulo",
    telefone="(11) 91234-5678",
    email="vitorvallio@vitorvallio.br"
)

# Chamadas aos métodos especiais:

# 1) __repr__ — mostrado no console interativo ou em repr(p)
print("repr(p):", repr(p))

# 2) __str__ — chamado por print(p)
print("\nprint(p):")
print(p)

# 3) __eq__ — vamos criar outra pessoa com mesmo e-mail
q = Pessoa("Outro Nome", "Rio", "(21) 99876-5432", "vitorvallio@vitorvallio.br")
print("\np == q ?", p == q)  # True, porque o email é igual

# 4) __len__ — retorna quantos “campos” essa Pessoa tem
print("\nLen(p):", len(p))  # 4
```

```python
# Exercício 3 - Crie a classe Smartphone com 2 atributos, tamanho e interface e crie a classe MP3Player com os 
# atributos capacidade. A classe MP3player deve herdar os atributos da classe Smartphone.

# Classe base Smartphone
class Smartphone:
    def __init__(self, tamanho, interface):
        """
        tamanho: string (ex: "6.5 polegadas")
        interface: string (ex: "Touchscreen", "Teclado físico")
        """
        self.tamanho = tamanho
        self.interface = interface

    def __str__(self):
        return f"Smartphone(tamanho={self.tamanho}, interface={self.interface})"


# Classe derivada MP3Player herda de Smartphone
class MP3Player(Smartphone):
    def __init__(self, tamanho, interface, capacidade):
        """
        capacidade: inteiro em MB ou GB (ex: 16, 32, 64)
        """
        # chama o construtor da classe-pai para inicializar tamanho e interface
        super().__init__(tamanho, interface)
        self.capacidade = capacidade

    def __str__(self):
        base = super().__str__()  # reusa a string do Smartphone
        return f"{base[:-1]}, capacidade={self.capacidade}MB)"


# Exemplo de uso:
if __name__ == "__main__":
    # Cria um MP3Player de 4 polegadas com interface "Botões" e 32MB de memória
    player = MP3Player(tamanho="4 polegadas", interface="Botões", capacidade=32)
    print(player)
    # Acesso aos atributos herdados e ao novo atributo
    print("Tamanho:", player.tamanho)
    print("Interface:", player.interface)
    print("Capacidade:", player.capacidade, "MB")
```
