## Polimorfismo de Classes em Python

```python
class Veiculo: 
    def __init__(self, marca, modelo):
        self.marca = marca
        self.modelo = modelo

    def acelerar(self):
        pass

    def frear(self):
        pass
```

```python
class Carro(Veiculo):
    def acelerar(self):
        print("O carro está acelerando.")

    def frear(self):
        print("O carro está freando.")
```

```python
class Moto(Veiculo):
    def acelerar(self):
        print("A moto está acelerando.")

    def frear(self):
        print("A moto está freando.")
```

```python
class Aviao(Veiculo):
    def acelerar(self):
        print("O avião está acelerando.")

    def frear(self):
        print("O avião está freando.")

    def decolar(self):
        print("O avião está decolando.")
```

```python
lista_veiculos = [Carro("Prosche", "911 Turbo"), Moto("Honda", "CB 1000R"), Aviao("Boeing", "757")]
```

```python
for item in lista_veiculos:
    item.acelerar()
    item.frear()
    if isinstance(item, Aviao):
        item.decolar()

    print("---")
```
