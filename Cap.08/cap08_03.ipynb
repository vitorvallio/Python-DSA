## Trabalhando com Metodos de Classes em Python

```python
class Circulo():
    pi = 3.14

    def __init__(self, raio = 5):
        self.raio = raio

    def area(self):
        return (self.raio * self.raio) * Circulo.pi

    def setRaio(self, novo_raio):
        self.raio = novo_raio

    def getRaio(self):
        return self.raio
```

```python
circ = Circulo()
```

```python
circ.getRaio()
```

```python
circ1 = Circulo(7)
```

```python
circ1.getRaio()
```

```python
print('O raio é:', circ.getRaio())
```

```python
print('Área igual a:', circ.area())
```

```python
circ.setRaio(3)
```

```python
print('Novo raio igual a:', circ.getRaio())
```
