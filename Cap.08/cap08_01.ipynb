## Classes

```python
class Livro():
    def __init__(self):
        self.titulo = 'Sapiens - Uma Breve História da Humanidade'
        self.isbn = 9988888
        print("Construtor chamado para criar um objeto desta classe.")
    
    def imprime(self):
        print("Foi criado o livro %s com ISBN %d" %(self.titulo, self.isbn))
```

```python
Livro1 = Livro()
```

```python
type(Livro1)
```

```python
Livro1.titulo
```

```python
Livro1.imprime()
```

```python
class Livro():
    def __init__(self, titulo, isbn):
        self.titulo = titulo
        self.isbn = isbn
        print("Construtor chamado para criar um objeto desta classe.")
    
    def imprime(self, titulo, isbn):
        print("Foi criado o livro %s com ISBN %d" %(self.titulo, self.isbn))
```

```python
Livro2 = Livro("The Communist Manifest", 140444785)
```

```python
Livro2.titulo
```

```python
Livro2.imprime("The Communist Manifest", 140444785)
```

```python
class Algoritmo():
    def __init__(self, tipo_algo):
        self.tipo = tipo_algo
        print("Construtor chamado para criar um objeto desta classe")
```

```python
algo1 = Algoritmo(tipo_algo = 'Random Forest')
```

```python
algo2 = Algoritmo(tipo_algo = 'Deep Learning')
```

```python
algo1.tipo
```

```python
algo2.tipo
```
