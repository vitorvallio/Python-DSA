```python
class Estudantes:
    def __init__(self, nome, idade, nota):
        self.nome = nome
        self.idade = idade
        self.nota = nota
```

```python
class Funcionarios:
    def __init__(self, nome, salario, cargo):
        self.nome = nome
        self.salario = salario
        self.cargo = cargo
    def listFunc(self):
        print("Funcionário(a) " + self.nome + " tem o salário de R$" + str(self.salario) + " e o cargo é " + self.cargo)
```

```python
Func1 = Funcionarios("Mary", 20000, "Ciêntista de Dados")
```

```python
Func1.listFunc()
```
